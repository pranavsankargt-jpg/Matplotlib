```python
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv(r"C:\Users\Ibscai48.DGVC-MCA\cars.csv")
print(df)
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
op=df['Mileage'].mean()
print(op)
```

```python
df['Mileage']=df['Mileage'].fillna(op)
print(df)
```

```python
plt.subplot(2,1,1)
x=df['Mileage']
y=df['Price']
plt.scatter(x,y,color="green",marker="o",s=80,label="Mileage vs Price")
plt.title("MILEAGE VS PRICE")
plt.xlabel("MILEAGE")
plt.ylabel("PRICE")
plt.legend()

plt.subplot(2,1,2)
plt.hist(df['Price'],color="red",edgecolor="green")
plt.title("PRICE DISTRIBUTION")
plt.xlabel("PRICE")
plt.ylabel("FREQUENCY")
plt.grid(True)
plt.show()
```
